# Automated File Sorter

```python
import os , shutil
```

```python
path = r"C:/Users/AKSHAT/Desktop/Testing Folder/"
```

```python
file_name = os.listdir(path)                           #listdir- (List Directory)
```

```python
Folder_names = ["csv_file","Image_File","Text_file"]

for loop in range (0,3):
    if not os.path.exists(path+Folder_names[loop]):
        print(path + Folder_names[loop])
        os.makedirs(path + Folder_names[loop])
```

```python
for file in file_name :
    if ".csv" in file and not os.path.exists(path+"csv_file/"+ file):        #(shutil.move = it moves the file from the directory)
        shutil.move(path + file, path + "csv_file/" + file)
    elif ".png" in file and not os.path.exists(path+"Image_file/"+ file):
        shutil.move(path + file, path + "Image_file/" + file)
    elif ".txt" in file and not os.path.exists(path+"Text_file/"+ file):
        shutil.move(path + file, path + "Text_file/" + file)
    
```
