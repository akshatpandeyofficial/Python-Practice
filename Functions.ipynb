# Multiple Parameters

```python

def squared(num,power):
    print(num ** power)
```

```python
squared(25,9)
```

# Arbitrary Arguments 

```python
args_tuple = (10,20,30,40,50)


def arg(*number):
    print(number[1]*number[2])
```

```python
arg(*args_tuple)
```

# Keyword Arguments 

```python
def squared(num,power):
    print(num ** power)
```

```python
squared(power=5,num=3)
```

# Keyword Arbitrary Argument

```python

def karg(**number):
    print('My name is:  ' + number['name'] + '   '+ 'My Roll Number is: '+ number ['roll_no'])
```

```python
karg( name = 'Akshat Pandey', roll_no = '52')
```
