# Group By And Aggregating Functions on A Dataset

```python
import pandas as pd
```

```python
df = pd.read_csv(r"C:\Users\AKSHAT\Desktop\J. N\Flavors.csv")
df
```

```python
group_by_frame = df.groupby('Base Flavor')
```

```python
group_by_frame.mean(numeric_only=True)
```

```python
df.groupby('Base Flavor').mean(numeric_only= True)
```

```python
df.groupby('Base Flavor').count()
```

```python
df.groupby('Base Flavor').min()
```

```python
df.groupby('Base Flavor').max()
```

```python
df.groupby('Base Flavor').sum()
```

```python
df.groupby('Base Flavor').agg({'Flavor Rating': ['mean','max','count','sum'], 'Texture Rating': ['mean','max','count','sum']})
```

```python
df.groupby(['Base Flavor','Liked']).mean(numeric_only = True).agg({'Flavor Rating': ['mean','max','count','sum']})
```

```python
df.groupby('Base Flavor').describe()
```
