```python
#For Loops
```

```python
integer = [1,2,3,4,5,6]
```

```python
for num in integer :
    print(num)
```

```python
for num in integer :
    print(num*num)
```

```python
dictionary = {'roll no' : 52, 'Name' : 'Akshat Pandey' , 'Stream' : 'Computer Science Btech' , 'favorite Hobby' : ['Football','singing','Travelling']}
```

```python
for value in dictionary.values():
    print(value)
    
```

```python
for key,value in dictionary.items():
    print(key , '-->', value)
```

```python
#Nested For Loops 
```

```python
shoes = ['sneakers','formal','casuals']
brand = ['nike','adidas','reebok','sparx']
```

```python
for one in shoes:
    for two in brand:
        print(two, 'has', one) 
```
