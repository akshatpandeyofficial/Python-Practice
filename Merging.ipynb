# Merging OF DataFrames


```python
import pandas as pd
```

```python
df1= pd.read_csv(r"C:\Users\AKSHAT\Desktop\J. N\LOTR.csv")
df1
```

```python
df2 = pd.read_csv(r"C:\Users\AKSHAT\Desktop\J. N\LOTR 2.csv")
df2
```
