# Joins


```python
import pandas as pd
```

```python
df = pd.read_csv(r"C:\Users\AKSHAT\Desktop\J. N\LOTR.csv")
df
```

```python
df2 = pd.read_csv(r"C:\Users\AKSHAT\Desktop\J. N\LOTR 2.csv")
df2
```

```python
df.join(df2, on = 'FellowshipID', how = 'outer', lsuffix= '_Left', rsuffix='_Right' )
```

```python
df4 = df.set_index('FellowshipID').join(df2.set_index('FellowshipID'),lsuffix= '_Left', rsuffix='_Right')
df4
```

```python
df4 = df.set_index('FellowshipID').join(df2.set_index('FellowshipID'),lsuffix= '_Left', rsuffix='_Right', how = 'outer')
df4
```
